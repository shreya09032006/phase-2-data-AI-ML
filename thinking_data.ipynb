```python
import json
```

```python
def load_data(filename):
    with open(filename , "r") as f:
        data =json.load(f)
    return data    
```

```python
load_data("data.json")
```

```python
def clean_data(data):
    text_to_num = {"one": 1, "two": 2, "three": 3, "four": 4, "five": 5}
    cleaned_data = []
    unique_users = set()

    for user in data:
        # 1. Clean the name
        name = user["name"].strip()
        user["name"] = name

        # 2. Remove duplicates
        if name in unique_users:
            continue
        unique_users.add(name)

        # 3. Clean the rating (text -> number)
        raw_rating = user["rating"].strip().lower()
        if raw_rating in text_to_num:
            user["rating"] = float(text_to_num[raw_rating])
        else:
            user["rating"] = float(raw_rating)

        # 4. Clean the age (missing -> None, text -> number)
        raw_age = user.get("age")
        if raw_age is None:
            user["age"] = None
        else:
            user["age"] = int(raw_age)

        # 5. Keep this user
        cleaned_data.append(user)

    return cleaned_data
        
```

```python
def get_insights(data):
    total_rating = 0
    for user in data:
        total_rating += float(user["rating"])
    print(f" avg rating = { total_rating/len(data) }" ) 

    poor_rating = 0
    for user in data:
        if(float(user["rating"])< 3):
          poor_rating += 1
    print(f"% of users with poor rating = {poor_rating / len(data) * 100}")   
```

```python
get_insights(data)
```

```python
#assignment3 
# Build recommendation feature
def get_recommendations(data):
    recommendations = []

    for user in data:
        curr_recomm = {}
        curr_recomm["name"] = user["name"]

        if float(user["rating"]) >= 4:
            curr_recomm["brand"] = "Apple"
        else:
            curr_recomm["brand"] = "Samsung"

        recommendations.append(curr_recomm)   # this line was missing

    return recommendations
            
```
